# Engagement 1: Duka Electronics Sales Exploration

## Executive Summary

- Duka Electronics recorded **1,000 orders** and generated **KES 71,792,500.00** in revenue during the six-month review period.
- **Nairobi** led the five branches with **KES 26,106,100.00**, while **Eldoret** recorded the lowest branch revenue.
- **Laptop** was the top product by revenue at **KES 41,535,000.00**, whereas **Tablet** led unit sales with **691 units**.
- The best month was **2026-04** at **KES 13,518,300.00**, compared with **2026-06** at **KES 10,903,400.00**.
- There were **83 Premium online orders** from Nairobi and Mombasa, contributing **KES 10,218,000.00**.

```python
import numpy as np
import pandas as pd
rng = np.random.default_rng(2026)

n = 1000

products = {"Phone": 15000, "Laptop": 65000, "Headphones": 2500,
            "Tablet": 28000, "Charger": 1200}



sales = pd.DataFrame({
    "order_id": range(1001, 1001 + n),
    "order_date": pd.to_datetime("2026-01-01")
                          + pd.to_timedelta(rng.integers(0, 180, n), unit="D"),
    "branch": rng.choice(["Nairobi", "Mombasa", "Kisumu", "Nakuru", "Eldoret"], n,
                         p=[0.4, 0.2, 0.15, 0.15, 0.1]),
    "product": rng.choice(list(products), n),
    "quantity": rng.integers(1, 6, n),
    "channel": rng.choice(["In-store", "Online", "Phone order"], n,
                          p=[0.55, 0.35, 0.10]),

})

sales["unit_price"] = sales["product"].map(products)

sales.to_csv("duka_sales.csv", index=False)
```

```python
sales = pd.read_csv("duka_sales.csv", parse_dates=["order_date"])
print("Shape:", sales.shape)
display(sales.dtypes.to_frame("dtype"))
display(sales.isna().sum().to_frame("missing_values"))
print("Date range: ", sales["order_date"].min().date(), "to", sales["order_date"].max().date())
for col in ["branch", "product", "channel"]:
  print(f"\n {col} categories:", sorted(sales[col].dropna().unique()))
  display(sales.head())
  display(sales.describe(include="all").T)
```

**Inspection finding:** The extract contains 1,000 orders. The date range and categories above are used as the baseline for the six-month review; no rows are removed during the sales exploration.

```python
sales["revenue"] = sales["quantity"] * sales["unit_price"]
sales["month"] = sales["order_date"].dt.to_period("M").astype(str)
sales["weekday"] = sales["order_date"].dt.day_name()
sales["is_weekend"] = sales["order_date"].dt.dayofweek >= 5
sales["price_band"] = np.where(sales["unit_price"] >= 28000, "Premium", "Value")
display(sales.head())
```

## Task 3 — Branch rankings

Revenue and order-count rankings answer different questions: revenue reflects value generated, while order count reflects transaction volume. A branch can lead revenue with fewer orders if its product mix has higher-priced items.

```python
branch_summary = sales.groupby("branch").agg(
    total_revenue = ("revenue", "sum"),
    orders = ("order_id", "nunique")
).sort_values("total_revenue", ascending=False)
branch_summary["revenue_rank"] = branch_summary["total_revenue"].rank(method="min", ascending=False).astype(int)
branch_summary["order_rank"] = branch_summary["orders"].rank(method="min", ascending=False).astype(int)
display(branch_summary)
print("Same ranking?\n", branch_summary["revenue_rank"].equals(branch_summary["order_rank"]))
```

## Task 4 — Top products by revenue and units

Revenue ranking favours products with higher price and/or quantity. Unit ranking favours products purchased most frequently in aggregate, regardless of price.

```python
product_summary = sales.groupby("product").agg(
    revenue = ("revenue", "sum"),
    units_sold = ("quantity", "sum")
)
display(product_summary.sort_values("revenue", ascending=False).head(3).rename_axis("Top 3 by revenue"))
display(product_summary.sort_values("units_sold", ascending=False).head(3).rename_axis("Top 3 by units sold"))
```

**Answer:** The two lists can differ because a high-priced product can generate substantial revenue from fewer units, while a low-priced product can lead unit sales without leading revenue.


```python
weekend_summary = sales.groupby("is_weekend").agg(
    orders = ("order_id", "nunique"),
    total_revenue=("revenue", "sum"),
    average_order_revenue = ("revenue", "mean")
).rename(index={False: "Weekday", True: "Weekend"})
display(weekend_summary)
```

**Answer:** Use order counts to compare activity, total revenue to compare overall contribution, and average order revenue to compare the value of a typical order. The measures should be interpreted together because a higher total may simply reflect more orders.

```python
monthly_revenue = sales.groupby("month")["revenue"].sum().sort_values()
best_month = monthly_revenue.idxmax()
worst_month = monthly_revenue.idxmin()
percentage_difference = (monthly_revenue.max() - monthly_revenue.min()) / monthly_revenue.min() * 100
display(monthly_revenue.sort_values(ascending=False).to_frame("revenue"))
print(f"Best: {best_month} ({monthly_revenue.max():,.2f})")
print(f"Worst: {worst_month} ({monthly_revenue.min():,.2f})")
print(f"Best month is {percentage_difference:.2f}% higher than worst month.")
```

**Answer:** Percentage difference is calculated relative to the worst month's revenue: (best − worst) ÷ worst × 100. This expresses how much higher the best month was than the lowest month.

```python
import os

premium_online_coast_capital = sales[
    (sales["price_band"] == "Premium") &
    (sales["channel"] == "Online") &
    (sales["branch"].isin(["Nairobi", "Mombasa"]))
].copy()

# Ensure the outputs directory exists
os.makedirs("outputs", exist_ok=True)

premium_online_coast_capital.to_csv("outputs/premium_online_coast_capital.csv", index=False)
print("Orders:", len(premium_online_coast_capital))
print("Total revenue:", premium_online_coast_capital["revenue"].sum())
display(premium_online_coast_capital.head())
```
